## Security notes
- HMAC-SHA256 is a symmetric message authentication code: sender and receiver share a secret. It is not a public-key signature.
- The included secret is for classroom demonstration only. In production, use a secret manager/environment variable, rotate keys, and protect the host-to-key mapping.
- The host allowlist authenticates a claimed host identity only when the sender can also prove possession of that host's secret. Network transport should additionally use TLS/mTLS.
- The example rejects commands before execution and never uses `eval` or `exec` on incoming data.
