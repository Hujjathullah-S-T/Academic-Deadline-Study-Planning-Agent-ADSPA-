## Academic Deadline & Study Planning Agent (ADSPA)

This notebook accompanies the runnable Python implementation. It demonstrates two distinct ADSPA agent instances negotiating academic study support using a **contract-net-inspired protocol**.

### Agents
- **ADSPA-Agent-A — Coordinator:** wants 120 minutes of support and prefers a lower price.
- **ADSPA-Agent-B — Study Support Provider:** offers available study time and balances price against workload.

### Protocol sequence
`CALL_FOR_PROPOSAL → PROPOSAL → COUNTER_PROPOSAL → REVISED_PROPOSAL → ACCEPTANCE → COMMIT_ACK → CONSENSUS_CONFIRMED`

Both agents calculate utilities. The agreement is finalized only after the provider's commitment acknowledgement and the coordinator's consensus confirmation.

## Utility model

**Coordinator utility**
- 65% weight on meeting the requested study-time allocation.
