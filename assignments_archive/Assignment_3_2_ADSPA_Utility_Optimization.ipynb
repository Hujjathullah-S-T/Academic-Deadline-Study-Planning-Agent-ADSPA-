# Assignment 3.2: Self-Interested Agent Utility Optimization for E-Commerce
## Academic Deadline & Study Planning Agent (ADSPA)

The agent selects one study-related product offer from 10 alternatives. It is self-interested: it maximizes a weighted utility score while respecting budget, delivery, quality, study-value, and stock constraints.

### Mathematical utility function
\[
U(o)=0.30Q + 0.20D + 0.20C + 0.30S - 0.04P_{total}
\]
where \(D=\mathrm{clip}(10(1-\text{delivery days}/10),0,10)\), and \(P_{total}=\text{price}+\text{shipping}\).

Hard constraints: total price ≤ 100 credits; delivery ≤ 5 days; quality ≥ 6; study value ≥ 6; stock ≥ 1.

## Algorithm
1. Build ten candidate offers.
2. Validate numeric ranges and hard constraints.
3. Compute utility for each feasible offer.
4. Reject infeasible offers instead of letting a high score override constraints.
5. Select the feasible offer with maximum utility using exhaustive search (`argmax`).
6. Run boundary tests for invalid quality, over-budget totals, unavailable stock, excessive delivery time, and negative price.
