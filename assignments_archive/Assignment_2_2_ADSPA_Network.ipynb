# Assignment 2.2: Distributed Agent Messaging using Sockets
## Academic Deadline & Study Planning Agent (ADSPA)

**Architecture:** TCP client-server over `127.0.0.1:50525`  
**Serialization:** UTF-8 JSON with newline message framing  
**Agent-A:** transfers its active state and sends a remote task-evaluation instruction  
**Agent-B:** acknowledges state, processes the instruction autonomously, and returns typed JSON responses.

### Required dual-terminal run
1. In Terminal 1, run `python Assignment_2_2_ADSPA_Network.py server`.
2. In Terminal 2, run `python Assignment_2_2_ADSPA_Network.py client`.
3. Both commands must run on the same computer for the default localhost setup. If using two computers, set the server host to a reachable interface and pass that IP to the client; ensure the port is allowed by the firewall.

## Message schema

Each newline-delimited JSON object contains:
- `protocol`: protocol version (`ADSPA/1.0`)
- `message_id`: unique UUID
- `message_type`: `STATE_TRANSFER`, `INSTRUCTION`, `PING`, or response type
- `sender`: agent identifier
- `timestamp`: timezone-aware timestamp
- `reply_to`: request message ID for responses
- `payload`: typed object containing state, instruction, or result
